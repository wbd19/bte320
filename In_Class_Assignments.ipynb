```python
sales = float(input("Enter sales ($): "))
rate = float(input("Enter tax rate (%): "))

cogs = 0.4 * sales
gross_profit = sales - cogs
taxes = rate * gross_profit
net = gross_profit - taxes

print(f"Sales: {sales}$")
print(f"Cost of Goods sold: {cogs}$")
print(f"Gross profit: {gross_profit}$")
print(f"Taxes: {taxes}$")
print(f"Net Income: {net}$")
```

Numerical Calculator

```python
num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /, //, **, %): ")
num2 = float(input("Enter the second number: "))


if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    result = num1 / num2
elif operator == "//":
    result = num1 // num2
elif operator == "**":
    result = num1 ** num2
elif operator == "%":
    result = num1 % num2
else:
    print("Error: Invalid operator.")
    result = None

print(f"Result: {result}")
```

Parking Garage (for/while loop)

```python
base_fee = 5
hourly_rate = 2.50

for hours_parked in range(1, 9):
    total_fee = base_fee + (hourly_rate * hours_parked)
    print(f"{hours_parked} {total_fee}")
```

```python
for hour in range(1, 9):
    fee = 5 + (2.5 * hour)
    if fee < 10:
      fee = 10
    elif fee > 20:
      fee = 20

    print(hour, fee)
```

```python
hour = 1
while hour < 9:
    fee = 5 + (2.5 * hour)
    if fee < 10:
      fee = 10
    elif fee > 20:
      fee = 20

    print(hour, fee)
    hour += 1
```

```python
from IPython.core.display import Math

ticket = 10
init_attendees = 20
fixed_cost = 200

print(f"Ad$ Profit (While Loop Version)")

ad_dollars = 0
while ad_dollars <= 200:
    revenue = (init_attendees + 2 * round(.5 ** (ad_dollars))) * 10
    costs = fixed_cost + ad_dollars
    profit = revenue - costs

    print(f"{ad_dollars}   {profit}")
    ad_dollars += 25
```

# **Rock Paper Scissors Game**

```python
import random

play = input("Do you want to play a game? (yes/no): ")

while play == "yes":
  p_move = input("Enter a choice (rock, paper, scissors): ")
  c_move = random.choice(["rock", "paper", "scissors"])

  print(f"Player move: {p_move}")
  print(f"Computer move: {c_move}")

  if p_move == c_move:
    print("Tie!")
  elif (
      (p_move == "rock" and c_move == "scissors") or
      (p_move == "paper" and c_move == "rock") or
      (p_move == "scissors" and c_move == "paper")
      ):
    print("You win!")
  else:
    print("You lose!")

  play = input("Do you want to play again? (yes/no): ")
```

# **Dynamic Dictionary**

```python
names = {}

while True:
  name = input("Give a name: ")
  if name == "":
    break

  if name in names:
    names[name] += 1
  else:
    names[name] = 1

  print(names)
```
