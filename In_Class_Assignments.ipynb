# **Numerical Calculator**

```python
num1 = float(input("Enter the first number: "))
operator = input("Enter an operator (+, -, *, /, //, **, %): ")
num2 = float(input("Enter the second number: "))


if operator == "+":
    result = num1 + num2
elif operator == "-":
    result = num1 - num2
elif operator == "*":
    result = num1 * num2
elif operator == "/":
    result = num1 / num2
elif operator == "//":
    result = num1 // num2
elif operator == "**":
    result = num1 ** num2
elif operator == "%":
    result = num1 % num2
else:
    print("Error: Invalid operator.")
    result = None

print(f"Result: {result}")
```

# **Rock Paper Scissors Game**

```python
import random

play = input("Do you want to play a game? (yes/no): ")

while play == "yes":
  p_move = input("Enter a choice (rock, paper, scissors): ")
  c_move = random.choice(["rock", "paper", "scissors"])

  print(f"Player move: {p_move}")
  print(f"Computer move: {c_move}")

  if p_move == c_move:
    print("Tie!")
  elif (
      (p_move == "rock" and c_move == "scissors") or
      (p_move == "paper" and c_move == "rock") or
      (p_move == "scissors" and c_move == "paper")
      ):
    print("You win!")
  else:
    print("You lose!")

  play = input("Do you want to play again? (yes/no): ")
```

# **Dynamic Dictionary**

```python
names = {}

while True:
  name = input("Give a name: ")
  if name == "":
    break

  if name in names:
    names[name] += 1
  else:
    names[name] = 1

  print(names)
```

# **Recursive Function: String reveal**

```python
def reverse(s):
  if len(s) < 2:
    return s
  else:
    return s[-1] + reverse(s[:-1])

reverse('abcde')
```

# **"Greeting" Class**

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal=False):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return f"Greeting object for {self.first} {self.last}"
```

```python
v = Greeting("Willem", "Davidson")
v.display()
v.display(formal=True)
print(v)
```

# **Student Class**

```python
class Student:
    def __init__(self, name, number):
        self.name = name
        self.number = number
        self.courses = []

    def get_name(self):
        return self.name

    def get_courses(self):
        return self.courses

    def set_name(self, new_name):
        self.name = new_name

    def __len__(self):
        return len(self.courses)

    def enroll(self, course_name):
        if course_name not in self.courses:
            self.courses.append(course_name)
            print(f"{self.name} enrolled in {course_name}.")
        else:
            print(f"{self.name} is already enrolled in {course_name}.")
```

```python
student1 = Student("John Doe", 1)
student1.enroll("History 101")
student1.enroll("Art 200")
print(f"Student name: {student1.get_name()}")
print(f"Student courses: {student1.get_courses()}")
print(f"Number of courses: {len(student1)}")
```
